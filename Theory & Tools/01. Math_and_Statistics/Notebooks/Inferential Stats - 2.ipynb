## 1. What is Hypothesis Testing?
Hypothesis testing is a probabilistic framework for making inferences about populations from samples. It quantifies uncertainty and helps data scientists avoid drawing conclusions from noise or randomness.


It’s a statistical method to check if what we see in sample data supports a claim (hypothesis) about the population.

- Formulate/Contruct Hypothesis on population mean
- Check whether the assumption all True or not

H0 = Null Hypothesis -- Status Quo

Ha - Alternate -- Research -- Challenge the Null

      - Mutually Exclusive (Both cannot be True at the same time)
      - Collectively Exhaustive (Either has to be True)

**Key Idea**: It's rooted in the scientific method. We formulate a testable statement (hypothesis) and use statistical evidence to either support or refute it. This prevents confirmation bias—e.g., seeing patterns where none exist.

**Fact**: Developed in the 1920s-1930s, it draws from probability distributions like the normal curve. In data science, it's crucial for model validation, feature selection, and A/B testing in platforms like Netflix or Google.

**Deeper Insight**: Hypotheses are probabilistic, not absolute. A rejected H₀ doesn't prove H₁; it just suggests H₁ is more likely. Always consider context—e.g., a p-value of 0.04 in a clinical trial might be life-saving, but in marketing, it could be trivial.


### Core Concepts
- **Null Hypothesis (H₀)**: Often the "no effect" or "no difference" claim. It's conservative to avoid false alarms.

- **Alternative Hypothesis (H₁)**: What you're hoping to prove. Can be directional (one-tailed) or non-directional (two-tailed).

- **Significance Level (α)**: Your tolerance for error. Lower α (e.g., 0.01) means stricter evidence needed. Fact: In fields like particle physics, α is 5σ (p < 3e-7) for discoveries.

- **P-Value**: Measures how extreme your data is under H₀. A small p-value indicates the data is unlikely if H₀ is true. Tip: Don't interpret p as the probability H₀ is true—that's a common myth.
- **Errors and Power**: Type I error is like a false alarm; Type II is missing a real effect. Power is the test's sensitivity. Fact: Increasing sample size boosts power but doesn't change α.

- **Effect Size**: Beyond p-value, it measures the magnitude of the difference (e.g., Cohen's d: small=0.2, medium=0.5, large=0.8). Tip: Always report effect size for practical relevance.

**Diagram: Hypothesis Testing Outcomes**

Reality vs. Decision

|                   | H₀ True           | H₀ False          |
|-------------------|-------------------|------------------|
| Reject H₀        | Type I Error (α)  | Correct (Power)   |
| Fail to Reject H₀| Correct           | Type II Error (β) |

**Deeper Dive**: In Bayesian terms, hypothesis testing is frequentist (focuses on long-run error rates). Bayesians use posterior probabilities instead. For data science, frequentist is common due to simplicity, but hybrid approaches (e.g., using priors) are emerging.

## 3. Types of Hypothesis Tests
- **Parametric vs. Non-Parametric**: Parametric tests (e.g., t-test) are powerful but assumption-heavy. Non-parametric (e.g., Wilcoxon) are robust to outliers.

- **Common Tests**:
    - **Z-Test**: For proportions or large n means. Formula: Z = (x̄ - μ) / (σ/√n).

    - **T-Test**: Adjusts for unknown σ. Degrees of freedom (df) = n-1.
    - **ANOVA**: Extends t-test to multiple groups; checks if at least one mean differs.
    - **Chi-Square**: For independence in tables. Expected = (row total * col total)/grand total.
    
- **Deeper Dive**: For correlations, Pearson assumes linearity; Spearman is rank-based. In machine learning, hypothesis testing validates features (e.g., t-test for feature importance).

## 4. Common Tests and When to Use Them
- **Z-Test**: For large samples (n > 30), known population variance. E.g., Test if a sample mean differs from population mean.

- **T-Test**: For small samples, unknown variance. Types: One-sample, Paired, Independent.

- **Chi-Square Test**: For categorical data, e.g., independence in contingency tables.

- **ANOVA**: Compare means across >2 groups.

- **Fact**: Z-test assumes normality and large n; t-test is more robust for small n. Power increases with sample size.

## 5. Tips and Tricks
- **Power Analysis**: Before collecting data, calculate n needed. Formula: n = (Z_α + Z_β)^2 * σ^2 / δ^2 (δ = effect size).

- **Multiple Testing**: If testing 10 hypotheses, α inflates. Adjust with Holm-Bonferroni.

- **Confidence Intervals (CI)**: Complement p-values. If CI excludes H₀ value, reject.

- **Trick**: Visualize distributions with boxplots to spot outliers before testing.

- **Common Pitfall**: Simpson's Paradox—effects reverse when grouping data.

- **Fact**: In data science, hypothesis testing informs decisions like deploying a model if accuracy > baseline.

## 6. Practical Example: One-Sample T-Test
Scenario: A coffee shop claims μ=100mg caffeine. Sample: n=20, x̄=105mg, s=10mg. α=0.05, two-tailed.

- **Theory**: The t-test statistic is t = (x̄ - μ) / (s/√n). It follows a t-distribution with df=19. Under H₀, t should be near 0; large |t| suggests difference. P-value is the probability of |t| ≥ observed if H₀ true.

- **Assumptions**: Normality (check with Shapiro-Wilk), independence, no outliers. Here, simulated normal data.

- **Practical**: In Colab, add visualizations and diagnostics. Effect size: Cohen's d = (x̄ - μ)/s ≈ 0.5 (medium effect).

- **Interpretation**: Rejecting H₀ means the sample mean is unlikely from a population with μ=100. But consider CI: If it includes 100, be cautious. In real-world, replicate or increase n.

- **Extensions**: For non-normal data, use Wilcoxon signed-rank test. In A/B testing, this is like comparing control vs. variant groups.

````python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns

# Set style
sns.set(style="whitegrid")
````

````python
# Sample data (simulate 20 cups: mean=105, sd=10)
np.random.seed(42)
sample = np.random.normal(105, 10, 20)

# Hypothesized mean
mu = 100

# Check assumptions: Normality test
stat, p_norm = stats.shapiro(sample)
print(f"Shapiro-Wilk Normality Test: Stat={stat:.3f}, P={p_norm:.3f}")
if p_norm > 0.05:
    print("Data appears normal.")
else:
    print("Data may not be normal; consider non-parametric test.")
````

````python
# Perform one-sample t-test
t_stat, p_value = stats.ttest_1samp(sample, mu)

# Output results
print(f"\nSample Mean: {np.mean(sample):.2f}, SD: {np.std(sample, ddof=1):.2f}")
print(f"T-Statistic: {t_stat:.3f}, DF: {len(sample)-1}")
print(f"P-Value: {p_value:.3f}, α: 0.05")
````

````python
# Decision
if p_value < 0.05:
    print("Reject H₀: Significant evidence that mean caffeine ≠ 100mg.")
else:
    print("Fail to Reject H₀: No significant evidence.")
````

````python
# Effect Size (Cohen's d)
d = (np.mean(sample) - mu) / np.std(sample, ddof=1)
print(f"Effect Size (Cohen's d): {d:.2f} (Medium effect)")
````

````python
# 95% Confidence Interval
ci = stats.t.interval(0.95, len(sample)-1, loc=np.mean(sample), scale=stats.sem(sample))
print(f"95% CI: ({ci[0]:.2f}, {ci[1]:.2f})")
````

````python
# Visualizations
plt.figure(figsize=(12, 4))

# Histogram with normal curve
plt.subplot(1, 3, 1)
plt.hist(sample, bins=10, alpha=0.7, color='blue', edgecolor='black')
plt.axvline(mu, color='red', linestyle='--', label=f'H₀ Mean: {mu}')
plt.axvline(np.mean(sample), color='green', linestyle='-', label=f'Sample Mean: {np.mean(sample):.2f}')
plt.title('Sample Distribution')
plt.legend()
````

````python
# Q-Q Plot for Normality
plt.subplot(1, 3, 2)
stats.probplot(sample, dist="norm", plot=plt)
plt.title('Q-Q Plot (Normality Check)')
````

````python
# Boxplot
plt.subplot(1, 3, 3)
plt.boxplot(sample, vert=False)
plt.axvline(mu, color='red', linestyle='--')
plt.title('Boxplot')

plt.tight_layout()
plt.show()
````

 p~0.8 (normal), t~2.5, p~0.02 → Reject H₀.

Deeper Practical Insights:

- **Why This Matters**: In data science, this test could validate a model's prediction (e.g., if a feature's mean differs). Always check for outliers—they can skew results.

- **Extensions**: Add power analysis: from statsmodels.stats.power import TTestIndPower; analysis = TTestIndPower(); analysis.solve_power(effect_size=d, alpha=0.05, power=0.8).

## **Statement 1** : Sedan cars cost on an average more than 75000 irrespective of the band.


## **Step 1 : Contruction of Hypothesis**

Option 1 :

$ H_0 : \mu >= 75000 $

$ H_a : \mu < 75000 $

Assuming that the Null is True

**Conclusion** :
- Sedan cars cost on an average is equals or more than 75000 irrespective of the band.
- Sedan cars cost on an average is less than 75000 irrespective of the band  

Option 2 :

$ H_0 : \mu <= 75000 $

$ H_a : \mu > 75000 $

Assuming that the Null is True

**Conclusion** :
- Sedan cars cost on an average more than 75000 irrespective of the band.
- Sedan cars do not cost on an average more than 75000 irrespective of the band.

## **Step 2 : Choose the level of significance**

at $\alpha = 0.05 $

## **Step 3 : Choose the Test**

- 1 Sample T-test
- One Tail - Right Tail Test

## **Step 4 : Performing the Test**

````python
sample = cars.loc[cars['Type'] == 'Sedan', 'MSRP']
mu0 = 750000
````

```python
stats.ttest_1samp(sample, popmean=mu0, alternative = ('greater'# for Right Tail
                                                      'less' # for left tail
                                                      'two-sided' # two tail test
))


````python
t_stats, p_value = stats.ttest_1samp(sample, popmean=mu0, alternative = 'greater')
````

````python
print(f"The test statistics is : {t_stats:7.2f}")
print(f"The p-value is         : {p_value:7.2f}")
````

## **Step 5 : Conclusion**

````python
def conclude(p_value, alpha=0.05) :
  if p_value < alpha:
    print('Fail to reject the null hypothesis')
  else:
    print('Reject the null hypothesis')
````

````python
conclude(p_value)
````

````python
# Calculate Trimmed Variance
var1 = stats.tvar(sample_1)
var2 = stats.tvar(sample_2)

# Calculate the F stats
F = var1/var2

# Get the numerator and denominator degrees of freedom
df1 = len(sample_1) - 1
df2 = len(sample_2) - 1

# for p value = 2 * min(left tail p value, right tail p value)
left_tail = stats.f.cdf(F, df1, df2)
right_tail = 1 - stats.f.cdf(F, df1, df2)
p_value = min(left_tail, right_tail) * 2

print(f"The test statistics is : {F:7.2f}")
print(f"The p-value is         : {p_value:7.2f}")

conclude(p_value)
# Rejecting means that the variance is not equal
````

## ANOVA : more than 2 samples

````python
# the price of Sedan cars for top 6 makes is same

top_6_makes = cars['Make'].value_counts()[:6].index
top_6_makes
````

````python
subdata = cars[cars['Make'].isin(top_6_makes)]
subdata = subdata[subdata['Type'] == 'Sedan' ]
````

````python
from statsmodels.formula.api import ols # Linear Regression
from statsmodels.stats.anova import anova_lm # ANOVA table
````

Formula ->

Variable ~ Treatment

**Variable** : the variable of interest | the one we're comparing | - Price

**Treatment** : categorical or the group column | - Make

````python
formula = 'MSRP ~ Make'

# Develop a linear model
model = ols(formula = formula, data = subdata).fit()

# get the anova results:
anova_results = anova_lm(model)

anova_results
````

**Conclusion** : Atleast one is different

Which one or more?

Multiple T-test

H0 : the pair in question has similar mean

Ha : the pair in question has different mean

````python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
````

````python
results = pairwise_tukeyhsd(subdata['MSRP'], subdata['Make'])
results.summary()
````

Reject = True | Null is rejected | pair is different

````python
results.plot_simultaneous()
````

## Chi - Square Test

types:
- Test of Independence

## **Statement** : Makers have preference for types

$ H_0 : $ Make and Type are Independent.

$ H_a : $ Make and Type are Dependent.

## **Step 2 : Choose the level of significance**

at $\alpha = 0.05$

## **Step 3 : Choose the Test**

- Chi Square Test for independence

## **Step 4 : Performing the Test**

````python
subdata = cars.loc[(cars['Make'].isin(top_6_makes)) & (cars['Type'] != 'Sedan') , ['Make','Type']]
````

````python
# Observation Table
obs = pd.crosstab(subdata['Make'], subdata['Type'])
obs
````

````python
test_stats, p_value, dof, expected_freq = stats.chi2_contingency(obs)
````

````python
conclude(p_value)
````

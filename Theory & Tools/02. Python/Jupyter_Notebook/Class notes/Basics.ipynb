### if elif else

grading :
- 90 and above--> A
- 75 - 90 --> B
- 55 - 75 --> C
- 35 - 55 --> D
- belo 35 --> F

````python
score= float(input("Please enter the score out of 100: "))
if score >= 90: 
    grade = 'A'
else : # (if score < 90)
    if score >= 75:
        grade  = 'B'
    else : # if score < 75
        if score >= 55:
            grade = 'C'
        else: # " if score < 55"
            if score >= 35:
                grade = 'D'
            else : # if score < 35
                grade = 'F'
````

````python
score= float(input("Please enter the score out of 100: "))
if score >= 90: 
    grade = 'A'
elif score >= 75:
    grade = 'B'
elif score >= 55:
    grade = 'C'
elif score >= 35:
    grade = 'D'
else :
    grade = 'F'
print(grade)
````

### Loop - else block
else block is executed when no break statement is encountered

````python
char = input("Enter single character: ")
for ch in 'Welcome to the Python class!':
    if ch == char:
        print(f"Character {char} encountered")
        break
else:
    print(f"{char} not present!")
print("Done")
````

### Argument Matching 
- Positional
- Keyword

````python
def example(first, second):
    print(f'first: {first}')
    print(f'Scnd : {second}')
````

````python
example("One", "Two") # positional matching
````

````python
example("Two", "One") # positional matching
````

````python
example(second = 'Two', first = "One") # keyword matching
````

````python
def welcome_message(name, greeting):
    print(f"Welcome {name}, {greeting}.")
````

````python
welcome_message('Happy to have you', 'Alan')
````

````python
welcome_message(greeting = 'Happy to have you', name = 'Alan')
````

````python
welcome_message('Alan')
````

````python
# default values 
def welcome_message_with_default(name, greeting= 'happy to have you!'):
    print(f"Welcome {name}, {greeting}")
````

````python
welcome_message_with_default('Alan')
````

````python
welcome_message_with_default('Alan', 'its our honour to have you here')
````
